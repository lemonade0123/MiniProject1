```python
def create_sample_data():
    # 일주일치 날짜 생성
    start_date = datetime(2025, 4, 10)
    dates = [(start_date + timedelta(days=i)).strftime('%Y-%m-%d') for i in range(7)]

    # 키워드 풀
    all_keywords = ['탄핵', '대통령', '경제', '부동산', '트럼프', '환율',
                   '관세', '선거', '정치', '주식', '물가', '금리']

    data = []
    for date in dates:
        # 각 날짜마다 랜덤으로 5개 키워드 선택 (실제 데이터는 이렇게 하지 않음)
        keywords = np.random.choice(all_keywords, 5, replace=False)
        counts = np.random.randint(10, 100, 5)  # 랜덤 등장 횟수

        for k, c in zip(keywords, counts):
            data.append({'date': date, 'keyword': k, 'count': c})

    return pd.DataFrame(data)

# 샘플 데이터 생성 (실제로는 아래 주석처럼 CSV 파일을 읽어오세요)
df = create_sample_data()
# df = pd.read_csv('weekly_keywords.csv')  # 실제 데이터를 읽을 때는 이 코드 사용

# 2. 각 날짜별로 상위 5개 키워드만 필터링
def get_top_5_per_day(df):
    result = pd.DataFrame()
    for date in df['date'].unique():
        day_data = df[df['date'] == date].sort_values('count', ascending=False).head(5)
        result = pd.concat([result, day_data])
    return result
df = get_top_5_per_day(df)

print("데이터 미리보기:")
print(df.head(10))

# 3. 모든 고유 키워드 목록 만들기 (일주일 동안 나타난 모든 키워드)
all_unique_keywords = df['keyword'].unique()
print(f"\n일주일 동안 나타난 모든 고유 키워드: {len(all_unique_keywords)}개")
```

```python
def visualize_heatmap(df):
    # 피벗 테이블 생성 (날짜 x 키워드, 값은 등장 횟수)
    pivot_df = df.pivot_table(index='date', columns='keyword', values='count', fill_value=0)

    # 상위 10개 키워드에 대한 히트맵 생성
    plt.figure(figsize=(14, 8))
    sns.heatmap(pivot_df, annot=True, fmt='g', cmap='YlGnBu')
    plt.title('일주일간 키워드 등장 횟수 히트맵')
    plt.ylabel('날짜')
    plt.xlabel('키워드')
    plt.tight_layout()
    plt.show()
```

```python
def make_keyword_ranking_table(df):
    # 날짜별로 groupby 후, 등장 횟수 내림차순 정렬
    ranking_table = {}

    for date in sorted(df['date'].unique()):
        # 해당 날짜의 데이터 중 count 기준 정렬 후 상위 5개 키워드
        keywords = df[df['date'] == date].sort_values('count', ascending=False)['keyword'].tolist()[:5]
        ranking_table[date] = keywords

    # 딕셔너리를 DataFrame으로 변환
    ranking_df = pd.DataFrame(ranking_table).T
    ranking_df.columns = [f'순위 {i+1}' for i in range(5)]
    ranking_df.index.name = '날짜'

    return ranking_df
```

```python
make_keyword_ranking_table(df)
```

```python
visualize_heatmap(df)
```
